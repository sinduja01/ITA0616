```python
import pandas as pd
import numpy as np
```

```python
data = pd.DataFrame([
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'no'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'yes']
])

data.columns = [
    'Sky', 'AirTemp', 'Humidity',
    'Wind', 'Water', 'Forecast', 'EnjoySport'
]

print(data)
```

```python
# Find-S Algorithm

num_attribute = len(data.columns) - 1

# Initialize the most specific hypothesis
hypothesis = ['0'] * num_attribute

print("Initial Hypothesis:", hypothesis)

# Apply Find-S algorithm
for index, row in data.iterrows():

    if row['EnjoySport'] == 'yes':

        for i in range(num_attribute):

            if hypothesis[i] == '0':
                hypothesis[i] = row.iloc[i]

            elif hypothesis[i] != row.iloc[i]:
                hypothesis[i] = '?'

    print("After Example", index + 1, ":", hypothesis)

# Display final hypothesis
print("\nFinal Maximally Specific Hypothesis:")
print(hypothesis)
```
