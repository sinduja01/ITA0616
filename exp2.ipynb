```python
import pandas as pd
import numpy as np
```

```python
data = pd.DataFrame([
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'no'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'yes']
])

data.columns = [
    'Sky', 'AirTemp', 'Humidity',
    'Wind', 'Water', 'Forecast', 'EnjoySport'
]

data.to_csv('enjoysport.csv', index=False)

print(data)
```

```python
data = pd.read_csv('enjoysport.csv')

print("Training Data:")
print(data)
```

```python
# Candidate Elimination Algorithm

num_attributes = len(data.columns) - 1

# Initialize Specific and General boundaries
S = [['0'] * num_attributes]
G = [['?'] * num_attributes]

# Function to check whether hypothesis covers an example
def covers(h, x):
    return all(h[i] == '?' or h[i] == x[i]
               for i in range(len(h)))

# Function to check generality
def more_general(h1, h2):
    return all(h1[i] == '?' or h1[i] == h2[i]
               for i in range(len(h1)))

# Process training examples
for index, row in data.iterrows():

    x = list(row.iloc[:-1])
    target = row.iloc[-1]

    if target == 'yes':

        # Remove inconsistent hypotheses from G
        G = [g for g in G if covers(g, x)]

        # Generalize S
        for i in range(num_attributes):
            if S[0][i] == '0':
                S[0][i] = x[i]
            elif S[0][i] != x[i]:
                S[0][i] = '?'

    else:

        # Remove inconsistent hypotheses from S
        S = [s for s in S if not covers(s, x)]

        # Specialize G
        new_G = []

        for g in G:
            if covers(g, x):

                for i in range(num_attributes):
                    if S and S[0][i] != '?' and S[0][i] != x[i]:
                        h = g.copy()
                        h[i] = S[0][i]
                        new_G.append(h)

            else:
                new_G.append(g)

        G = new_G

    print("After Example", index + 1)
    print("S Boundary:", S)
    print("G Boundary:", G)
    print()

print("Final Specific Boundary (S):")
print(S)

print("\nFinal General Boundary (G):")
print(G)
```
